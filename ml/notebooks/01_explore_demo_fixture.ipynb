# 01 · Explore the demo fixture

⚠️ `fixtures/scenario-kubu-raya.json` is **synthetic**. Use it to check wiring and plots only —
never to report model performance.

```python
import matplotlib.pyplot as plt

from pavana_ml.data.fixtures import load_scenario, scenario_frames

scenario = load_scenario()
hourly, hotspots = scenario_frames(scenario)
print(scenario["meta"]["title"], "|", scenario["meta"]["description"][:80], "…")
hourly.head()
```

```python
local = hourly["valid_at"].dt.tz_convert(scenario["location"]["timezone"])
fig, ax = plt.subplots(figsize=(9, 3))
ax.plot(local, hourly["pm25_ug_m3"], marker="o")
ax.set(title="PM2.5 (SYNTHETIC demo)", ylabel="µg/m³", xlabel="local time (WIB)")
ax.grid(alpha=0.3)
```

```python
loc = scenario["location"]
fig, ax = plt.subplots(figsize=(5, 5))
ax.scatter(
    hotspots["lon"],
    hotspots["lat"],
    s=hotspots["frp_mw"].fillna(1) * 3,
    c="tab:red",
    alpha=0.6,
    label="hotspot (synthetic)",
)
ax.scatter([loc["lon"]], [loc["lat"]], marker="*", s=250, c="tab:orange", label=loc["name"])
ax.set(title="Synthetic hotspots (size ∝ FRP)", xlabel="lon", ylabel="lat")
ax.legend()
ax.set_aspect("equal")
```
