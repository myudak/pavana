# 00 · Setup & conventions

Read this before touching data. These rules are shared with the API (`packages/shared`).

| Thing | Convention |
|---|---|
| Time | Store **UTC** (`datetime64[ns, UTC]`). Convert to WIB/WITA/WIT only for display. |
| PM2.5 | `µg/m³`, column suffix `_ug_m3`. Missing = `NaN`, **never 0**. |
| Wind speed | `m/s` (`_ms`). Convert km/h at the loader. |
| Wind direction | Meteorological **FROM** degrees (`wind_dir_from_deg`). Convert to flow **TO** before alignment math. |
| Provenance | Every source keeps `provider`, `issued_at` (forecasts), `valid_at`, `fetched_at`. |
| Hotspots | Satellite detections, not confirmed fires. No detection ≠ no fire. |

## Wind: FROM vs TO
A westerly wind (`from 270°`) moves air **toward the east** (`to 90°`).

```python
from pavana_ml.features.wind import wind_from_to_flow_bearing, wind_from_to_flow_vector

for frm in (0, 90, 135, 270):
    e, n = wind_from_to_flow_vector(frm, 1.0)
    to = float(wind_from_to_flow_bearing(frm))
    print(f"from {frm:>3}° → flows to {to:>5.0f}°  (east={float(e):+.2f}, north={float(n):+.2f})")
```
